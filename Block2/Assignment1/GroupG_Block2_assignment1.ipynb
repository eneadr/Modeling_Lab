### Classical thermodynamics

#### 1. Basic assumptions of the Langmuir model

The Langmuir adsorption model is based on three main assumptions:

- All adsorption sites are energetically equivalent, meaning that a molecule has the same adsorption energy regardless of the site on which it adsorbs.

- Each adsorption site can only adsorb one molecule. Once all sites are occupied, the surface reaches its saturation loading $q_{sat}$.

- There are no interactions between adsorbed molecules.

#### 2a. Behaviour at high pressure

The Langmuir equation is

$$q=\frac{q_{sat}kP}{1+kP}.$$

At high pressure, $P\rightarrow\infty$ and therefore $kP\gg1$.

We can divide both the numerator and denominator by $kP$:

$$q=\frac{q_{sat}}{\frac{1}{kP}+1}.$$

When $P\rightarrow\infty$, $\frac{1}{kP}\rightarrow0$. Therefore,

$$\lim_{P\rightarrow\infty}q=q_{sat}.$$

Thus, at sufficiently high pressure, all adsorption sites become occupied and the gas uptake reaches a plateau corresponding to the saturation loading $q_{sat}$.

#### 2b. Behaviour at low pressure

At low pressure, $P\rightarrow0$ and $kP\ll1$. The denominator of the Langmuir equation can then be approximated by

$$1+kP\approx1.$$

Therefore,

$$q\approx q_{sat}kP.$$

Hence, at low pressure, gas uptake is directly proportional to pressure. This is the Henry-law regime.

#### 3. Temperature dependence of the Henry coefficient

From the low-pressure limit, $q\approx q_{sat}k(T)P$, and from Henry's law, $q=K_H(T)P$.

Therefore,

$$\boxed{K_H(T)=q_{sat}k(T)}.$$

The Van't Hoff equation is

$$k(T)=k_0\exp\left(-\frac{\Delta H}{RT}\right).$$

Substituting this expression into the Henry coefficient gives

$$\boxed{K_H(T)=q_{sat}k_0\exp\left(-\frac{\Delta H}{RT}\right)}.$$

Taking the natural logarithm,

$$\ln K_H=\ln(q_{sat}k_0)-\frac{\Delta H}{RT}.$$

Therefore, the Henry coefficient has the same exponential temperature dependence as the Langmuir equilibrium constant.

For an exothermic adsorption process, $\Delta H<0$. In this case, increasing temperature decreases $K_H$, meaning that adsorption becomes less favourable at higher temperature.

#### 4. Determination of the Langmuir parameters

##### Step 1 – Fit the adsorption isotherms to the Langmuir model

For each temperature, the adsorption data consist of values of gas uptake $q$ measured at different pressures $P$.

The Langmuir model is

$$q(P)=\frac{q_{sat}k(T)P}{1+k(T)P}.$$

At 263 K, both $q_{sat}$ and $k(263)$ can be obtained by fitting this equation to the adsorption isotherm.

Once $q_{sat}$ has been determined, it is assumed to be independent of temperature, as required in the assignment.

Therefore:

- At 263 K: fit $q_{sat}$ and $k(263)$.
- At 303 K: keep $q_{sat}$ fixed and fit only $k(303)$.
- At 343 K: keep $q_{sat}$ fixed and fit only $k(343)$.

The shape of the fitted curves provides information about the parameters:

- $q_{sat}$ determines the high-pressure plateau.
- $q_{sat}k(T)$ determines the initial slope in the low-pressure region.

This first step therefore separates the saturation capacity of the material from its temperature-dependent adsorption affinity.

#### Fitted Langmuir parameters used below

After fitting the isotherms, the following values are used:

- $q_{sat}=4.72\ \mathrm{mmol\,g^{-1}}$
- $k(263)=6.75\times10^{-6}\ \mathrm{Pa^{-1}}$
- $k(303)=1.48\times10^{-6}\ \mathrm{Pa^{-1}}$
- $k(343)=5.16\times10^{-7}\ \mathrm{Pa^{-1}}$

```
qsat = 4.72

temperatures = np.array([263, 303, 343])

k_values = np.array([
    6.75e-6,
    1.48e-6,
    5.16e-7
])

def langmuir(P, qsat, k):
    # Langmuir adsorption model
    return qsat * k * P / (1 + k * P)

# Pressure range used to visualize the fitted isotherms
P = np.linspace(0, 5e6, 500)

plt.figure(figsize=(8, 5))

for T, k in zip(temperatures, k_values):
    q = langmuir(P, qsat, k)
    plt.plot(P, q, label=f"{T} K")

plt.xlabel("Pressure, P (Pa)")
plt.ylabel("Gas uptake, q (mmol/g)")
plt.title("Langmuir adsorption isotherms")
plt.legend()
plt.grid()

plt.show()
```

To plot the Langmuir adsorption isotherms, we use the fitted Langmuir parameters in the Langmuir equation and evaluate the gas uptake over a common pressure range for each temperature.

The adsorption curves increase rapidly at low pressure and progressively approach the saturation loading $q_{sat}$.

All curves have the same theoretical saturation loading because $q_{sat}$ is assumed to be independent of temperature.

At a given pressure, adsorption is greater at lower temperature because the Langmuir equilibrium constant $k(T)$ is larger.

The decrease of adsorption with increasing temperature is consistent with an exothermic adsorption process.

##### Step 2 – Determine $k_0$ and $\Delta H$ using the Van't Hoff equation

The Van't Hoff equation is

$$k(T)=k_0\exp\left(-\frac{\Delta H}{RT}\right).$$

Taking the natural logarithm gives

$$\boxed{\ln k=\ln k_0-\frac{\Delta H}{R}\frac{1}{T}}.$$

This equation has the form of a straight line, $y=b+mx$, with

$$y=\ln k,\qquad x=\frac{1}{T},\qquad m=-\frac{\Delta H}{R},\qquad b=\ln k_0.$$

Therefore,

$$\boxed{\Delta H=-mR}$$

and

$$\boxed{k_0=e^b}.$$

A linear fit of $\ln k$ versus $1/T$ therefore provides both the heat of adsorption and the standard equilibrium constant.

#### 5.a Complete the code

The fitted values of $k(T)$ obtained above are now used to perform the Van't Hoff linear fit and calculate the remaining parameters $k_0$ and $\Delta H$.

```
T = np.array([263, 303, 343])

k = np.array([
    6.75e-6,
    1.48e-6,
    5.16e-7
])

# Linearized Van't Hoff variables
inverse_T = 1 / T
ln_k = np.log(k)

# Fit ln(k) = slope * (1/T) + intercept
slope, intercept = np.polyfit(inverse_T, ln_k, 1)

R = 8.314e-3  # kJ/(mol.K)
delta_H = -slope * R
k0 = np.exp(intercept)

print("Slope =", slope)
print("Intercept =", intercept)
print("Delta H =", delta_H, "kJ/mol")
print("k0 =", k0, "1/Pa")
```

```
# Values predicted by the linear Van't Hoff fit
ln_k_fit = slope * inverse_T + intercept

plt.figure(figsize=(8, 5))

plt.scatter(inverse_T, ln_k, label="Data")
plt.plot(inverse_T, ln_k_fit, label="Linear fit")

plt.xlabel("1/T (1/K)")
plt.ylabel("ln(k)")
plt.title("Van't Hoff plot")
plt.legend()
plt.grid()

plt.show()
```

The plot of $\ln k$ versus $1/T$ is approximately linear, as predicted by the Van't Hoff equation.

From the slope,

$$\Delta H\approx-24.2\ \mathrm{kJ\,mol^{-1}}.$$

The negative value of $\Delta H$ confirms that adsorption is exothermic.

From the intercept,

$$k_0\approx1.06\times10^{-10}\ \mathrm{Pa^{-1}}.$$

The fact that $k(T)$ decreases as temperature increases is consistent with the negative adsorption enthalpy.

#### 5.b Calculate Henry's coefficients for the different temperatures

At low pressure, $K_H=q_{sat}k(T)$. We therefore calculate one Henry coefficient for each temperature using the fitted Langmuir parameters.

```
qsat = 4.72

T = np.array([263, 303, 343])

k = np.array([
    6.75e-6,
    1.48e-6,
    5.16e-7
])

# Henry coefficient in the low-pressure limit
KH = qsat * k

for temperature, value in zip(T, KH):
    print(
        f"Henry coefficient at {temperature} K = "
        f"{value:.5e} mmol/(g.Pa)"
    )
```

```
plt.figure(figsize=(8, 5))

plt.plot(T, KH, marker="o")

plt.xlabel("Temperature (K)")
plt.ylabel("Henry coefficient, KH (mmol/(g.Pa))")
plt.title("Temperature dependence of Henry's coefficient")

plt.grid()

plt.show()
```

The Henry coefficient decreases strongly when temperature increases.

This behaviour follows directly from

$$K_H(T)=q_{sat}k_0\exp\left(-\frac{\Delta H}{RT}\right).$$

Because $\Delta H$ is negative, adsorption is exothermic. Increasing the temperature therefore makes adsorption less favourable, resulting in a smaller equilibrium constant $k$ and consequently a smaller Henry coefficient.

Physically, this means that at low pressure the MOF has a greater affinity for CO$_2$ at 263 K than at 303 K or 343 K.

### Statistical Thermodynamics

#### 1. Canonical Ensemble

The canonical ensemble describes a system with a fixed number of particles ($N$) and fixed volume ($V$), held at constant temperature ($T$) by contact with a large heat bath (thermal reservoir). Energy can be exchanged with the reservoir, so the instantaneous energy of the system fluctuates while $N$, $V$ and $T$ remain fixed.

In adsorption studies, this ensemble is useful for calculating equilibrium average-energy properties at a fixed number of adsorbed particles.

#### 2. Partition function

For a system in the canonical ensemble, the canonical partition function is defined as the sum over all accessible microstates $v$, weighted by the Boltzmann factor:

$$Q(N,V,T)=\sum_v e^{-\beta E_v},\qquad \beta=\frac{1}{k_BT}.$$

The partition function contains the statistical information needed to obtain macroscopic thermodynamic properties.

#### 3. Probability

The probability of finding the system in a specific microstate $v$ with energy $E_v$ is given by the Boltzmann distribution, normalized by $Q$:

$$P_v=\frac{e^{-\beta E_v}}{Q}
=\frac{e^{-E_v/k_BT}}{\sum_v e^{-E_v/k_BT}}.$$

#### 4. Proofs

##### 4.1 Entropy

We want to prove

$$\langle S\rangle
=
k_B\ln Q
+
k_BT\left(\frac{\partial\ln Q}{\partial T}\right)_{N,V}.$$

Start from the statistical definition of entropy:

$$S=-k_B\sum_v P_v\ln P_v.$$

Since

$$\ln P_v=-\beta E_v-\ln Q,$$

we obtain

$$
S
=
-k_B\sum_vP_v(-\beta E_v-\ln Q)
=
k_B\beta\sum_vP_vE_v
+
k_B\ln Q\sum_vP_v.
$$

Using $\sum_vP_v=1$ and $\sum_vP_vE_v=\langle E\rangle$:

$$
S
=
\frac{\langle E\rangle}{T}
+
k_B\ln Q.
$$

From the partition function,

$$
\frac{\partial\ln Q}{\partial T}
=
\frac{1}{Q}\frac{\partial Q}{\partial T}
=
\frac{\langle E\rangle}{k_BT^2},
$$

so

$$
\langle E\rangle
=
k_BT^2
\left(\frac{\partial\ln Q}{\partial T}\right)_{N,V}.
$$

Substituting this result into the entropy expression gives

$$
\boxed{
\langle S\rangle
=
k_B\ln Q
+
k_BT
\left(\frac{\partial\ln Q}{\partial T}\right)_{N,V}
}.
$$

##### 4.2 Average Energy

Starting directly from the partition function,

$$Q=\sum_v e^{-\beta E_v},$$

we differentiate with respect to $T$:

$$
\frac{\partial Q}{\partial T}
=
\sum_v e^{-\beta E_v}\frac{E_v}{k_BT^2}.
$$

Dividing by $Q$ gives

$$
\frac{\partial\ln Q}{\partial T}
=
\frac{1}{k_BT^2}
\sum_vP_vE_v
=
\frac{\langle E\rangle}{k_BT^2}.
$$

Therefore,

$$
\boxed{
\langle E\rangle
=
k_BT^2
\left(\frac{\partial\ln Q}{\partial T}\right)_{V,N}
}.
$$

##### 4.3 Pressure

We want to prove

$$
P
=
k_BT
\left(\frac{\partial\ln Q}{\partial V}\right)_{N,T}.
$$

From

$$
dA=-S\,dT-P\,dV+\mu\,dN,
$$

at constant $T$ and $N$,

$$
P
=
-\left(\frac{\partial A}{\partial V}\right)_{T,N}.
$$

Using the canonical relation

$$
A=-k_BT\ln Q,
$$

we obtain

$$
\boxed{
P
=
k_BT
\left(\frac{\partial\ln Q}{\partial V}\right)_{N,T}
}.
$$

##### 4.4 Chemical Potential

We want to prove

$$
\mu
=
-k_BT
\left(\frac{\partial\ln Q}{\partial N}\right)_{V,T}.
$$

From

$$
dA=-S\,dT-P\,dV+\mu\,dN,
$$

at constant $T$ and $V$,

$$
\mu
=
\left(\frac{\partial A}{\partial N}\right)_{T,V}.
$$

Again using

$$
A=-k_BT\ln Q,
$$

we obtain

$$
\boxed{
\mu
=
-k_BT
\left(\frac{\partial\ln Q}{\partial N}\right)_{T,V}
}.
$$

The main idea is that the partition function $Q$ connects the microscopic energy levels of the system to macroscopic thermodynamic quantities through derivatives of $\ln Q$.
