Adsorption Using AiiDAlab

1. AiiDAlab offers for adsorption studies 3 types of calculations :
Pore analysis and the pore volume, to understand the geometry of the MOF, isotherm calculations to simulate the quantity of gas adsorbed at various pressure a ta gien temperature, and Henry coefficient.

2. Computation of :
- the density accessible surface area : 0.576983 g/cm3
- accessible surface area : 3964.65 A^2
- accessible probeoccupiable volume : 13737.4 A^3
- porosity : 0.77498

3. The Henry coefficients of CO2 and CH4 are respectively 4.95112e-6 and 1.12113e-6 mol/kg/Pa. Because CO2 has the bigger Henry coefficient, it will better be absorbed by the IRMOF-1 at low pressure, meaning that this material can be used to separate the CO2 and CH4 and mainly recover the CO2.
 


4. CO2 absorption isotherm at T= 300K :

```python
import matplotlib.pyplot as plt

pressure = [
    0.2, 0.8, 1.4, 2.0, 2.6, 3.2, 3.8, 4.4, 5.0,
    5.6, 6.2, 6.8, 7.4, 8.0, 8.6, 9.2, 9.8, 10.2
]

loading_CO2 = [
    0.09649599694768,
    0.3948153845504,
    0.69862192675748,
    1.02791613976,
    1.4028933499715,
    1.6763744341876,
    2.166452017609,
    2.50301914559,
    2.9011138355549,
    3.2198882965348,
    3.655126802283,
    3.9100359786074,
    4.4598553636384,
    5.0184412086882,
    5.4556602850332,
    5.9664851571657,
    6.7251411008668,
    7.0717084869433
]

plt.figure(figsize=(7,5))

plt.plot(pressure, loading_CO2, marker='o', linewidth=2)

plt.xlabel("Pressure (bar)")
plt.ylabel("CO$_2$ loading (mol kg$^{-1}$)")
plt.title("CO$_2$ adsorption isotherm at 300 K")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```
