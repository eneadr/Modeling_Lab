## Question 4 — $CO_2$ working capacity

The $CO_2$ working capacity, which measures how much $CO_2$ can be theoretically recovered, is calculated from the **binary $CO_2$ loadings predicted by IAST (Ideal Adsorbed Solution Theory)**:

$$
WC = q_{\mathrm{adsorption}} - q_{\mathrm{desorption}}
$$

where $q_{\mathrm{adsorption}}$ and $q_{\mathrm{desorption}}$ are the $CO_2$ loadings in the binary $CO_2/N_2$ mixture at the adsorption and desorption pressures, respectively.

The flue gas composition is assumed to be:

$$
y_{\mathrm{CO_2}} = 0.15, \qquad y_{\mathrm{N_2}} = 0.85
$$

The adsorption and desorption pressures are:

$$
P_{\mathrm{ads}} = 1.0\ \mathrm{bar}, \qquad
P_{\mathrm{des}} = 0.2\ \mathrm{bar}
$$

IAST uses the pure-component $CO_2$ and $N_2$ isotherms obtained in Question 3 to predict the adsorption of the binary mixture.

Since the simulated pure-component isotherms start at $0.2$ bar, the physical point $(P,q)=(0,0)$ is added before interpolation. This allows IAST to evaluate the low partial pressures encountered in the binary mixture.

```
def make_iast_isotherm(dataset):
    pressure = np.array(dataset["pressure"], dtype=float)
    loading = np.array(dataset["loading"], dtype=float)

    # Add the physical origin
    pressure = np.insert(pressure, 0, 0.0)
    loading = np.insert(loading, 0, 0.0)

    df = pd.DataFrame({
        "Pressure": pressure,
        "Loading": loading
    })

    return pyiast.InterpolatorIsotherm(
        df,
        pressure_key="Pressure",
        loading_key="Loading"
    )
```

```
y_CO2 = 0.15
y_N2 = 0.85

P_ads = 1.0  # bar
P_des = 0.2  # bar

gas_composition = np.array([y_CO2, y_N2])
```

```
iast_results = []

for material, datasets in isotherms.items():

    # IAST requires both pure-component isotherms
    if datasets["CO2"] is None or datasets["N2"] is None:
        print(f"{material}: skipped — incomplete CO2/N2 isotherm data")
        continue

    co2_iso = make_iast_isotherm(datasets["CO2"])
    n2_iso = make_iast_isotherm(datasets["N2"])

    # Binary mixture loadings at adsorption pressure
    q_ads = pyiast.iast(
        P_ads * gas_composition,
        [co2_iso, n2_iso]
    )

    # Binary mixture loadings at desorption pressure
    q_des = pyiast.iast(
        P_des * gas_composition,
        [co2_iso, n2_iso]
    )

    q_CO2_ads = q_ads[0]
    q_CO2_des = q_des[0]

    working_capacity = q_CO2_ads - q_CO2_des

    iast_results.append({
        "Material": material,
        "q_CO2_ads (mmol/g)": q_CO2_ads,
        "q_CO2_des (mmol/g)": q_CO2_des,
        "Working capacity (mmol/g)": working_capacity
    })

working_capacity_df = pd.DataFrame(iast_results)

working_capacity_df
```

## Question 5 — $CO_2/N_2$ selectivity

The $CO_2/N_2$ selectivity is calculated from the binary loadings predicted by IAST:

$$
S_{\mathrm{CO_2/N_2}}
=
\frac{q_{\mathrm{CO_2}}}{q_{\mathrm{N_2}}}
\frac{y_{\mathrm{N_2}}}{y_{\mathrm{CO_2}}}
$$

where $q_{\mathrm{CO_2}}$ and $q_{\mathrm{N_2}}$ are the equilibrium loadings of each component in the binary mixture, and $y_{\mathrm{CO_2}}$ and $y_{\mathrm{N_2}}$ are their gas-phase mole fractions.

For the simulated flue gas:

$$
y_{\mathrm{CO_2}} = 0.15, \qquad y_{\mathrm{N_2}} = 0.85
$$

The selectivity is evaluated at the adsorption pressure of $1.0$ bar and at $298$ K.

```
selectivity_results = []

for material, datasets in isotherms.items():

    # IAST requires both pure-component isotherms
    if datasets["CO2"] is None or datasets["N2"] is None:
        print(f"{material}: skipped — incomplete CO2/N2 isotherm data")
        continue

    co2_iso = make_iast_isotherm(datasets["CO2"])
    n2_iso = make_iast_isotherm(datasets["N2"])

    # Binary loadings at the adsorption pressure
    q_mix = pyiast.iast(
        P_ads * gas_composition,
        [co2_iso, n2_iso]
    )

    q_CO2 = q_mix[0]
    q_N2 = q_mix[1]

    selectivity = (q_CO2 / q_N2) * (y_N2 / y_CO2)

    selectivity_results.append({
        "Material": material,
        "q_CO2 (mmol/g)": q_CO2,
        "q_N2 (mmol/g)": q_N2,
        "CO2/N2 selectivity": selectivity
    })

selectivity_df = pd.DataFrame(selectivity_results)

selectivity_df
```

```
# Question 5 — selectivity expression
y_CO2 = 0.15
y_N2 = 0.85

def co2_n2_selectivity(q_CO2, q_N2):
    return (q_CO2 / q_N2) * (y_N2 / y_CO2)
```
